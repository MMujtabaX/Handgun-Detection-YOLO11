# CV - Object Detection
## Model: YOLO | Dataset: Custom

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import os
import glob as glob

import cv2
from PIL import Image
from ultralytics import YOLO
```

# 1. Train on Custom Dataset

```python
!pwd
```

```python
ROOT_DIR = '/mnt/c/Users/AI/Documents/sandbox/OD'
```

```python
model = YOLO("yolo_models/yolo11n.pt")
```

```python
results = model.train(data=os.path.join(ROOT_DIR, 'dataset', 'data.yaml'), batch=8, epochs=10, workers=4)
```

## 1.1 Validation

```python
metrics = model.val()
```

```python
print(metrics.box.map)
print(metrics.box.map50)
print(metrics.box.map75)
print(metrics.box.maps)
```

## 1.2 Test Model

```python
test_images = sorted(glob.glob(os.path.join(ROOT_DIR, 'dataset', 'test', 'images') + '/*.jpg'))
len(test_images)
```

```python
results = model.predict(test_images, save=True, imgsz=640, conf=0.5)
```
